`````python
# Run this cell first. It finds your three gold tables from 01_build_sdp
# and prints the dashboard prompt with your schema filled in.

TARGET_CATALOG = "genie_workshop_catalog"
GOLD_TABLES = ["gold_sales_velocity", "gold_inventory_health", "gold_supply_chain_coverage"]

PROMPT_TEMPLATE = """Create a dashboard called "Crisp Retail Analytics" from these gold tables:
- <target_schema>.gold_sales_velocity
- <target_schema>.gold_inventory_health
- <target_schema>.gold_supply_chain_coverage

Page 1, "Sales overview":
- Counters for total revenue, total units sold and number of products sold
- Line chart of weekly revenue
- Bar chart of revenue by retailer
- Bar chart of the top 10 products by revenue
- Bar chart of revenue by store state

Page 2, "Inventory and supply":
- Bar chart of product count by stock status
- Table of fast-selling products that are out of stock or low on stock
- Bar chart of product count by supply status for each retailer

Add filters for retailer, store state and a week date range that apply to every chart on both pages."""

user = spark.sql("SELECT current_user()").first()[0]
schema_name = user.split("@")[0].replace(".", "_")   # first.last becomes first_last
target_schema = f"{TARGET_CATALOG}.{schema_name}"

missing = []
for table in GOLD_TABLES:
    try:
        found = spark.catalog.tableExists(f"{target_schema}.{table}")
    except Exception:
        found = False
    if not found:
        missing.append(table)

print(f"Target schema:  {target_schema}")
if missing:
    print("\nGold table(s) NOT FOUND: " + ", ".join(missing))
    print("Finish 01_build_sdp first, then run its Check your output cell.")
else:
    print("\nOK: all three gold tables exist.")

print("\nCopy everything between the lines into Genie Code:\n" + "-" * 72)
print(PROMPT_TEMPLATE.replace("<target_schema>", target_schema))
print("-" * 72)
`````
