## Business Context

You've just joined **NovaBrand**, a fast-growing CPG company that sells snacks, beverages and household goods through major US retailers. The data team has ingested harmonized point-of-sale and inventory data from **Crisp**, but the Sales Ops team is drowning in data requests.

## What is Genie One?

Genie One is **the single front door for business users** on Databricks: one place to ask questions, find assets and act on the answers.

| Capability | What It Does | Exercise |
|---|---|---|
| **Discovery** | Ask a question to find the dashboards and agents that already exist | 1 |
| **Chat** | Ask data questions in plain language; Genie One answers from the most relevant governed sources and shows them | 2 |
| **Dashboards** | Surface a dashboard from Chat and explain what it shows | 3 |
| **Skills** | Encode business conventions once (metric rules, key markets) | 4 |
| **Memory** | Teach Genie One how you like your answers | 5 |
| **Agent Creation** | Turn a conversation into a draft Genie Agent the team can use | 6 *(optional)* |
| **Scheduled Tasks** | Get recurring checks delivered to your inbox | 7 |
| **Documents** | Draft shareable documents with cited sources, straight from a conversation | 8 |

> **Key distinction:** Genie One is the experience. **Genie Agents** are the domain-specific setups (tables, instructions, metrics) that power many of its answers behind the scenes. You use the agent built in Module 2 throughout this lab, and in the optional Exercise 6 you see how a business user can draft a new one from a conversation.

## Exercise 3: Surface a Dashboard and Explain a Trend from Chat

### The Scenario

The VP needs to reference an existing dashboard during a meeting but does not remember where it is stored in the workspace.

> "Pull up the retail dashboard, I don't remember where it is. And explain why Texas dipped in November 2022."

The task is to surface the dashboard directly from Chat, without browsing the workspace.

### What to Do

**3.1 Start a new conversation in Genie One Chat**

**3.2 Find the dashboard**
Ask:

> Show me the retail sales dashboard

Genie One should return a link to the dashboard. Click it to open the dashboard without navigating the workspace.

**3.3 Ask a follow-up that builds on the dashboard**
Ask:

> Why did Texas sales drop in November 2022?

**3.4 Check that you can trust the answer**
Look at the sources listed under the response and open the query behind the answer. Does the explanation match what the dashboard shows for Texas in November 2022?

**3.5 (Optional) Get something ready for the meeting**
Ask:

> Summarise this in three bullet points I can share with my team

Check that the summary keeps the key numbers and the link to the dashboard.

### What You Should See

- The dashboard is found from a plain-language request, with a direct link.
- The follow-up answer is built on the same governed data the dashboard uses, with sources you can check.
- The answer turns into a short summary ready to share in the meeting.

### Before You Move On

| # | Check | Done? |
|---|---|---|
| 1 | Genie One returned a **link to the dashboard** and I opened it from Chat | |
| 2 | My Texas follow-up was answered **without leaving Chat** | |
| 3 | I checked the **sources and query** behind the answer, and they match the dashboard | |

**Where the answer comes from:** Genie One picks the most relevant governed sources for each question, and can combine several of them:

```
User question
    │
    ├── Looking for an asset?  ("Show me the retail sales dashboard")
    │       → Link to the dashboard or Genie Agent
    │
    ├── Asking about the data? ("Why did Texas sales drop?")
    │       → Answer built from a Genie Agent, dashboard data, tables or metric views
    │
    └── Every answer → Sources listed underneath, so you can check them
```

**Key takeaway:** From one conversation, the VP finds the dashboard, gets an explanation they can check, and has a summary ready for the meeting, all without browsing the workspace.

## Exercise 4: Encode Your Business Conventions with a User Skill

### The Scenario

After several Chat sessions, the VP identifies a recurring issue: Genie One does not automatically include both units and revenue when comparing retailers, and it lacks awareness of NovaBrand's key market conventions.

> "Every time I compare retailers, I have to remind it to show units AND revenue. And it doesn't know that California, Texas, and Florida are our top 3 states. Oh, and Q4 is our holiday season, it should always stand out. Can we encode our conventions so it applies them automatically?"

This task introduces user skills: personal domain knowledge files that Genie One loads when your question matches them.

### What to Do

**4.1 Create the skill from Chat**
Type:

> Create a user skill called crisp-retail that knows the following: We are a CPG brand analyzing sell-through data from the Crisp retail data platform. Data covers 4 retailers (Retailer 1, 4, 5, 7), 5 suppliers, and 2,288 products across all US states. Date range is 2021–2022. Sales amounts are in USD. When comparing retailers, always include units sold alongside revenue. California, Texas, and Florida are our top 3 states by volume.

**4.2 Verify the skill is registered**
Open **Customization** and confirm `crisp-retail` appears under your skills.

**4.3 Test both conventions**
Start a new conversation and ask:

> Compare our retailers in our key states

Confirm the answer shows both **units sold** and **revenue**, and focuses on **California, Texas and Florida**, without you asking for either.

**4.4 Update the skill**
Type:

> Update my crisp-retail skill to also know that Q4 (Oct–Dec) is our critical holiday season and all quarterly comparisons should highlight Q4 performance separately.

**4.5 Test the update**
Start a new conversation and ask:

> How did we perform by quarter in 2022?

Confirm the answer calls out **Q4 holiday season** performance separately.

---

### Before You Move On

| # | Check | Done? |
|---|---|---|
| 1 | The skill `crisp-retail` appears in **Customization** | |
| 2 | In a new conversation, "Compare our retailers in our key states" returned **units and revenue** for **California, Texas and Florida**, without me asking | |
| 3 | After updating the skill, the quarterly question called out **Q4 holiday season** performance separately | |

**Key takeaway:** The VP sets their conventions once and Genie One applies them in every new conversation. Skills are created and updated in plain language, with no code, YAML or admin involved.

> **What you covered:** Creating a skill from Chat, testing it in a new conversation, and updating it as conventions evolve. **Next → Task 5.**


## Exercise 5: Memory, Teach Genie One Your Preferences

### The Scenario

The skill fixed the business rules. But the VP reads answers on her phone between meetings, and she keeps asking for the same layout in every new conversation.

> "The numbers are right now, but I read these between meetings. Give me the headline first, keep it short, and always show the change vs last year. Can Genie One remember how I like my answers?"

This task introduces memory: personal preferences that Genie One applies in every new conversation.

### What to Do

**5.1 Save a memory**
Tell Genie One:

> Remember that I like short answers: start with a one-line headline, then at most 3 bullets, and always show the change vs the previous year.

**5.2 Check that it applies**
Start a new conversation and ask:

> What were total sales in December 2022?

Check that the answer starts with a headline, has at most 3 bullets and compares with December 2021. It should still show **units and revenue**: your skill and your memory work together.

**5.3 Trigger a suggested memory**
Mention something about yourself in passing:

> I'm preparing my quarterly business review with Retailer 4. How did they do in Q4 2022?

Watch for Genie One to offer to remember that Retailer 4 is your key account. **Confirm** or **Reject** it.

**5.4 Update a memory**
Tell Genie One:

> Actually, also show the change vs the previous quarter. Update my memory.

**5.5 Spot a citation**
Ask any sales question, then find the **memory citation** in the response. It shows which preference shaped the answer.

**5.6 Delete a memory**
Tell Genie One:

> Forget that Retailer 4 is my key account.

Then ask *"How are our retailers doing?"* and check that Retailer 4 no longer gets special focus.

### Before You Move On

| # | Check | Done? |
|---|---|---|
| 1 | In a new conversation, the December 2022 answer used my format (headline, ≤ 3 bullets, change vs last year) | |
| 2 | I confirmed or rejected a suggested memory | |
| 3 | I updated a memory, and the next answer included the change vs the previous quarter | |
| 4 | I found a memory citation in at least one response | |
| 5 | I deleted a memory, and it no longer shaped the answer | |

**Reference: memory features**

| Feature | How it works |
|---|---|
| Explicit save | Ask Genie One to remember something → saved straight away |
| Suggested memories | Mention something about yourself or correct how Genie answers → it offers to remember it; you confirm or reject |
| Citations | When a memory shapes an answer, it is cited → you can always see why |
| Update / delete | Ask in plain language → "Update my memory about..." or "Forget that..." |
| Privacy | Memories are private to you, not shared with other users |
| Skill vs memory | Skill = business rules about the data (Task 4). Memory = your personal preferences and context |

**Key takeaway:** Skills teach Genie One the business; memory teaches it about the VP. She sets her preferences once, and every answer comes in the format she needs between meetings.

> **What you covered:** Explicit memory saves, suggested memories, memory citations, updating and deleting memories in plain language, the privacy model, and how memory and skills work together. **Next → Task 6.**

## (Optional) Exercise 6: Create a Genie Agent, From Conversation to Governed Asset

### The Scenario

The VP wants to scale beyond individual use. She has 12 regional managers who ask similar questions every week.

> "This works for me. But can you turn what we've been doing into something the whole team can use, without me being the bottleneck?"

This task turns a conversation into a governed, reusable Genie Agent that the whole team can use.

### Prerequisites

To create a Genie Agent from Genie One, you need the **Workspace access** or **Databricks SQL access** entitlement. Users with only Consumer access cannot create agents.

### What to Do

**6.1 Build the context**
In one thread, ask the questions a regional manager would ask every week:

> What are total sales by retailer for Q4 2022?

> Which products are driving Retailer 5's sales in California?

> How does average selling price compare across retailers?

**6.2 Create the agent**
Under the last answer, click the overflow menu **(•••) → Create Agent**. Name it `NovaBrand Retail Performance`.
*Alternative:* ask *"Save this analysis as a Genie Agent so my sales team can reuse it."*

**6.3 Review what was carried over**
Open the agent and check its tables, sample questions and instructions.
Your skill and memories are **personal**, so the team won't get them. Add the team conventions to the agent's instructions:

> When comparing retailers, always show units sold and revenue. California, Texas and Florida are our top 3 states. Highlight Q4 (Oct–Dec) separately in quarterly comparisons.

**6.4 Share it with the team**
Share the agent with your lab neighbour (in real life: the regional managers' group).

**6.5 Test it as a team member**
Your neighbour asks in Genie One:

> Use the NovaBrand Retail Performance agent: compare our retailers

Check that the answer shows **units and revenue**, without them asking for it.

### Before You Move On

| # | Check | Done? |

| 2 | I added the team conventions to the agent's instructions | |

**Key takeaway:** A business user turns a conversation into a first draft of an agent in minutes. Conventions move from personal (skill, memory) to team-wide (agent instructions), and the data team reviews and certifies the agent before a wider rollout (as you did in Module 2).


## Exercise 7: Scheduled Tasks, Automate Recurring Insights

---

### The Scenario

The VP needs a recurring weekly check: did any state's sales drop more than 20% versus the same week last year? She wants it delivered automatically every Monday before her 8:30 standup.

> *"I need this every Monday before my standup. And I don't want to ask someone every time."*

This task introduces **scheduled tasks**: recurring prompts that Genie One runs for you and delivers by email.


---

### What to Do

- [ ] **7.1 Create the scheduled task** by asking Genie One in Chat:
  > *"Every Monday at 8 AM, check whether any state had a sales decline of more than 20% in the most recent week of data, compared with the same week the year before. Send me the states that dropped, with both units and revenue."*

  *Did Task 6?* Point the task at your agent instead: *"Every Monday at 8 AM, using the NovaBrand Retail Performance agent, check whether..."*

- [ ] **7.2 Confirm it was created:** the task appears in the **Schedules** section of the sidebar, set to Monday 8 AM in your time zone.
- [ ] **7.3 Test it:** click **Run now** to trigger a run straight away. A run can take a few minutes; continue reading the reference below while it runs.
- [ ] **7.4 Review the results:** open the new chat thread created by the run. Check that it lists the states that dropped, with units and revenue.
- [ ] **7.5 Edit the task in plain language:** ask Genie One:
  > *"Update my Monday sales alert: if no state dropped more than 20%, just send me a one-line all clear."*

  Click **Run now** again and check the change. No need to recreate the task.

---

### Before You Move On

| # | Check | Done? |
|---|---|---|
| 1 | My scheduled task appears in the **Schedules** sidebar section | |
| 2 | I ran it with **Run now** and got results in a new chat thread | |
| 3 | The results show the states that dropped, with **units and revenue** | |
| 4 | I updated the task from Chat, and the next run reflected the change | |

**Reference: what happens when a task runs**

- Genie One runs your prompt as a full chat session, on its own, with your permissions
- Results post to a **new chat thread** you can review or continue
- You receive an **email with the results**, including visualizations and a PDF attachment
- Each run is independent: open any past run to review it

**Key takeaway:** The VP gets her Monday check in her inbox before the standup, set up and changed in plain language. No jobs to configure, no tickets to file.

> **What you covered:** Creating a scheduled task in plain language, test runs, reviewing results in the generated thread, and updating the task from Chat. **Next → Task 8.**

## Exercise 8: Documents, From Insight to Shareable Artifact

> *"Turn our retailer analysis into a one-pager for the board. I want to send it tonight, without leaving the platform."*

This task introduces **documents**: shareable artifacts that Genie One drafts straight from a data conversation, with sources cited.

---

### What to Do

- [ ] **8.1 Pick up the analysis:** open the **conversation history** in the sidebar and select your retailer comparison thread from **Task 4** (or Task 6).
